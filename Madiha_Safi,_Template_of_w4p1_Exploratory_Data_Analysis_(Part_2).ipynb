# Part E — GroupBy Analysis

## E1. Conceptual model of `groupby()`

`groupby()` answers two connected questions:

1. **How should the rows be divided?**  
   This is determined by the grouping column(s).

2. **What should be calculated for each group?**  
   This is determined by the selected target variable and aggregation method.

A reusable pattern is:

```python
df.groupby(GROUPING_COLUMN)[TARGET_COLUMN].AGGREGATION()
```

For example:

```python
df.groupby("sex")["salary"].mean()
```

# Part H — Consolidated EDA Workflow

The following sequence collects the principal operations demonstrated across the four lecture sections.

````python
# 1. Load data
df = pd.read_csv("Salaries.csv")

# 2. Basic inspection
display(df.head())
print("Shape:", df.shape)
print("Columns:", list(df.columns))
display(df.dtypes)
display(df.values)

# 3. Descriptive statistics
display(df.describe())
print("Mean salary:", df["salary"].mean())
print("Maximum salary:", df["salary"].max())
print("Maximum years of service:", df["yrs.service"].max())

# 4. Column selection
display(df["rank"])
display(df[["yrs.service", "salary"]])
display(df[["yrs.service", "salary", "yrs.since.phd"]].min())

# 5. Categorical exploration
print("Unique rank values:", df["rank"].unique())
display(df["rank"].value_counts())

# 6. Grouped analysis
display(df.groupby("rank").first())
display(df.groupby("rank")["salary"].mean())
display(df.groupby("sex")["salary"].mean())
display(df.groupby("sex")["yrs.service"].mean())
display(df.groupby(["sex", "rank", "discipline"]).mean(numeric_only=True))

# 7. Conditional filtering + aggregation
df_prof_10 = df[
    (df["rank"] == "Prof") &
    (df["yrs.service"] < 10)
]
print("Prof, service < 10 — count:", len(df_prof_10))
print("Prof, service < 10 — mean salary:", df_prof_10["salary"].mean())

df_prof_more_10 = df[
    (df["rank"] == "Prof") &
    (df["yrs.service"] > 10)
]
print("Prof, service > 10 — count:", len(df_prof_more_10))
print("Prof, service > 10 — mean salary:", df_prof_more_10["salary"].mean())
````
