## 1. Why Do We Need Variables?

A variable gives a meaningful name to a value. Instead of repeating a literal value throughout a program, we store it once and refer to it by name. This makes the code easier to read, update, and reuse.


### Comparing Repeated Values with Variables

First, observe a version that repeats the literal value `25`. If that value changes, every occurrence must be edited manually.


````python
#Manipulating values without variables

print(25)
print(25+5)
print(25 * 2)
````

Now store the value once in `age`. Updating `age` automatically affects every later expression that uses the variable.


````python
# Manipulating values with variables
age = 25
age = 25

print(age)
print(age + 5)
print(age * 2)
````

## 2. Declaring and Initializing Variables

Python does not require a separate declaration statement. A variable is created when a value is assigned to a name.

```python
variable_name = value
```

The `=` symbol is the **assignment operator**: it associates the name on the left with the value on the right.


### Examples from Different Problem Domains

The following variables represent student information, a bank balance, weather data, and a game state. Notice that the names communicate the purpose of each stored value.


````python
# Examples of Declaring and Initialization of Variables

# Student information
name = "Alice"
age = 22
height = 1.65
graduated = False

# Bank account
account_holder = "John Smith"
account_balance = 1500.75
account_number = 100234
is_active = True

# Weather
city = "Bangkok"
temperature = 32.5
is_raining = False

# Game
player_name = "Hero"
level = 3
score = 250
game_over = False

print(name, age, height, graduated)
print(account_holder, account_balance, account_number, is_active)
print(city, temperature, is_raining)
print(player_name, level, score, game_over)
````

### 3.1 Python Uses Dynamic Typing

Python is **dynamically typed**. A variable name may refer to values of different types at different moments during execution. The type belongs to the current value, not permanently to the variable name.


### Example 6 — Reassignment and Case Sensitivity

This cell demonstrates two ideas: a variable can be reassigned to a value of another type, and Python treats `Name` and `name` as different identifiers.


````python
# Example 6: Dynamic type changing.
Name = "Alice"
name = "Bob"

print(type(15))
print(type(name))
print(Name)
print(name)
````

## 4. Assigning One Value to Multiple Variables

**Chain assignment** assigns the same value to several variable names in one statement.

```python
x = y = z = 100
```

This is useful when several settings or counters must begin with the same value.


### Example 7 — Chain Assignment

All three names receive the value `100`. Print them separately to verify the assignment.


````python
# Example - 7: Assigning a Single Value to Multiple Variables
x = y = z = 100

print(x)
print(y)
print(z)
````

### Example 8 — Practical Chain Assignment

Chain assignment is convenient when several scores or counters begin at the same value.


````python
# Example 8: Real-life Examples

#Initialize multiple players scores.
player1_score = player2_score = 0

print("player1_score =", player1_score)
print("player2_score =", player2_score)

#Initialize multiple quizes values.
quiz1 = quiz2 = quiz3 = 0

print("\nquiz1 = ", quiz1)
print("quiz2 = ", quiz2)
print("quiz3 = ", quiz3)
````

## 5. Assigning Different Values to Multiple Variables

Python supports **multiple assignment**, also called unpacking. The number of variables on the left must match the number of values on the right.

```python
name, age, grade = "John", 20, 90
```


### Example 9 — Unpacking Related Values

Values are assigned from left to right: `"John"` to `name`, `20` to `age`, and `90` to `grade`.


````python
# Example 9: Assigning Different Values to Multiple Variables
name, age, grade = "John", 20, 90

print(name)
print(age)
print(grade)
````

````python
# Example 10: Assigning Different Values to Multiple Variables
x, y = 5, 10
print(x)
print(y)
````

### Interpreting Standard Deviation

Standard deviation measures how widely values are distributed around their mean:

- A **small** standard deviation indicates that values tend to be close to the mean.
- A **large** standard deviation indicates greater variation.

Its unit is the same as the original variable, which makes it easier to interpret alongside the mean.


````python
# Calculate Standard Deviation
# of a particular Column
df['salary'].std()
````

### 6.10 Detecting Missing Values


````python
#Remove Rows with Missing Values: dropna()
#Checking the missing values
df.isnull().sum()
````

### 6.11 Removing Rows with Missing Values


````python
#Create a DataFrame without rows
#containing missing values:
df_clean = df.dropna()

#Compare the dimensions:
print("Before:", df.shape)
print("After:", df_clean.shape)
````

### Interpreting the Missing-Value Result

The dimensions remain unchanged because this dataset does not contain missing values in the inspected rows. In datasets that contain missing values, `dropna()` removes affected rows by default, so the number of rows may decrease.
